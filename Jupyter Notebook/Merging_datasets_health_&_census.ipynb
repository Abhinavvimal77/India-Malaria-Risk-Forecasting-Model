```python
import pandas as pd

# Define the file names
census_file = 'C:\\abhi\\project\\disease predict\\census-cleaned-2.csv'
health_file = 'C:\\abhi\\project\\disease predict\\District_Wise_Rural_HealthCare_Infrastructure(cleaned).csv'
output_file = 'merged_census_health_data.csv'

# Load the datasets
try:
    df_census = pd.read_csv(census_file)
    df_health = pd.read_csv(health_file)

    # --- Data Cleaning ---
    
    # 1. Drop unnecessary empty columns from the health dataset
    #    errors='ignore' prevents an error if the columns don't exist
    df_health_cleaned = df_health.drop(columns=['Unnamed: 6', 'Unnamed: 7'], errors='ignore')

    # 2. Clean the 'District' columns in both dataframes
    #    .str.strip() removes any leading or trailing whitespace
    df_census['District_cleaned'] = df_census['District'].str.strip()
    df_health_cleaned['District_cleaned'] = df_health_cleaned['District'].str.strip()

    # --- Merging ---
    
    # 3. Perform an inner merge on the cleaned 'District' columns
    #    'how='inner'' ensures only districts present in BOTH files are kept.
    merged_df = pd.merge(
        df_census, 
        df_health_cleaned, 
        on='District_cleaned',  # The common column to merge on
        how='inner'
    )

    # --- Post-Merge Cleanup ---
    
    # 4. The merge creates 'District_x' (from census) and 'District_y' (from health)
    #    We can drop 'District_y' and the 'District_cleaned' key
    merged_df = merged_df.drop(columns=['District_y', 'District_cleaned'], errors='ignore')
    
    # 5. Rename 'District_x' back to 'District' for a clean final file
    merged_df = merged_df.rename(columns={'District_x': 'District'})

    # --- Saving the File ---
    
    # 6. Save the merged dataframe to a new CSV file
    merged_df.to_csv(output_file, index=False)

    print(f"Successfully merged data and saved to: {output_file}")
    print(f"Original census rows: {len(df_census)}")
    print(f"Original health rows: {len(df_health_cleaned)}")
    print(f"Merged rows (matching districts): {len(merged_df)}")

except FileNotFoundError as e:
    print(f"Error: File not found. Make sure {e.filename} is in the same directory.")
except Exception as e:
    print(f"An error occurred: {e}")
```
